## Kamus Data (Data Dictionary)

Berdasarkan sumber dataset, berikut adalah penjelasan untuk 11 kolom (1 variabel waktu + 10 variabel operasional) yang terdapat di dalam dataset kualitas udara ini:

| Nama Kolom | Jenis Variabel | Satuan/Rentang | Keterangan |
| :--- | :--- | :--- | :--- |
| **`datetime`** | Waktu (*Datetime*) | ISO 8601 | Stempel waktu (*timestamp*) pencatatan data per jam berdasarkan waktu lokal IST (UTC+05:30). |
| **`pm25`** | Numerik kontinu | µg/m³ (15–600) | Partikulat halus berukuran kurang dari 2,5 mikrometer ($PM_{2.5}$). |
| **`pm10`** | Numerik kontinu | µg/m³ (30–900) | Partikulat kasar berukuran kurang dari 10 mikrometer ($PM_{10}$). |
| **`no2`** | Numerik kontinu | µg/m³ (5–80) | Nitrogen dioksida ($NO_2$), polutan yang sering menjadi proksi emisi kendaraan/lalu lintas. |
| **`co`** | Numerik kontinu | mg/m³ (0.05–4) | Karbon monoksida ($CO$), gas beracun hasil pembakaran tidak sempurna. |
| **`so2`** | Numerik kontinu | µg/m³ (1–20) | Sulfur dioksida ($SO_2$), biasanya dihasilkan dari aktivitas industri atau kembang api. |
| **`o3`** | Numerik kontinu | µg/m³ (5–120) | Ozon permukaan ($O_3$), polutan sekunder yang meningkat saat siang hari (reaksi fotokimia). |
| **`temp`** | Numerik kontinu | °C (12–45) | Suhu udara bola kering (*dry-bulb temperature*). |
| **`rh`** | Numerik kontinu | % (20–100) | Kelembapan relatif (*relative humidity*). |
| **`wind`** | Numerik kontinu | m/s (0.1–30) | Kecepatan angin pada ketinggian 10 meter. |
| **`rain`** | Numerik kontinu | mm/jam (0–150) | Curah hujan per jam. |
